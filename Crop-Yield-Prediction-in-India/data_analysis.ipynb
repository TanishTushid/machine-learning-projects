```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
data = pd.read_csv("crop_production.csv")
data 
```

```python
data.shape
```

```python
data.columns 
```

```python
data.describe()
```

```python
print(data["Crop"].unique())
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data = data.dropna()
```

```python
data.isnull().values.any()
```

```python
data.State_Name.unique()
```

```python
# Adding a new column Yield which indicates Production per unit Area. 
data["Yield"] = (data["Production"] / data["Area"])
data.head(10)
```

```python
sns.barplot(x = data["State_Name"], y = data["Production"])
plt.xticks(rotation = 90)
plt.show
```

```python
sns.barplot(x = data["State_Name"], y = data["Yield"])
plt.xticks(rotation = 90)
plt.show
```

```python
sns.barplot(x = data["Crop_Year"], y = data["Production"])
plt.xticks(rotation = 90)
plt.show
```

```python
sns.barplot(x = data["Crop_Year"], y = data["Yield"])
plt.xticks(rotation = 90)
plt.show
```

```python
sns.lineplot(x = data["Season"], y = data["Yield"])
plt.xticks(rotation = 90)
plt.show
```

```python
sns.lineplot(data=data, x="Season", y="Production")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(16, 8))
sns.barplot(data=data, x="Crop", y="Yield", palette="tab20", hue="Season", legend=False)
plt.xticks(rotation=60)
```

```python
sns.jointplot(x="Area",y="Production",data=data,kind="reg")
plt.xticks(rotation=90)
```

Conclusions Obtained from the dataset:

Kerela is the largest producer of crop in India.
Production per unit area (Yield ) of Puducherry is maximum.
In the year 2011, crop yield was maximum.
